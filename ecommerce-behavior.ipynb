```python
sample_chunks = []

chunk_size = 500000

chunk_reader =pd.read_csv("Data/raw/2019-Oct.csv", chunksize=chunk_size)

for i, chunk in enumerate(chunk_reader):
    print(f"Processing chunk {i}: shape is {chunk.shape}")
    chunk_sample = chunk.sample(n=10000, random_state=42)
    sample_chunks.append(chunk_sample)


ecommerce_sample = pd.concat(sample_chunks, ignore_index = True)
print("Final sample shape:", ecommerce_sample.shape)
```

```python
ecommerce_sample.to_csv("Data/raw/sample_oct_850k.csv", index=False)
```
